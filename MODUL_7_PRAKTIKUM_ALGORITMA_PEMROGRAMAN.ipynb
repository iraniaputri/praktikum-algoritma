```python
def cek_prima(angka):
    if angka < 2:
        return False
    for i in range(2, angka):
        if angka % i == 0:
            return False
    return True

def tampilkan_hasil(angka, status):
    if status:
        print(f"{angka} adalah bilangan Prima")
    else:
        print(f"{angka} bukanlah bilangan Prima")

angka = int(input("Masukkan angka: "))
status = cek_prima(angka)
tampilkan_hasil(angka, status)
```

```python
def cari_akhiran(angka):
    if angka == 1:
        return "st"
    elif angka == 2:
        return "nd"
    elif angka == 3:
        return "rd"
    else:
        return "th"

print("Ordinal Number - ketik 0 untuk menghentikan program")

while True:
    angka = int(input("Masukkan angka: "))
    akhiran = cari_akhiran(angka)
    print(angka, akhiran)

    if angka == 0:
        print("Terima kasih telah menggunakan program saya")
        break
```
